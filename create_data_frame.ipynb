```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import avg, max, min

data = [
    (1, "Alice", "IT", 50000),
    (2, "Bob", "HR", 40000),
    (3, "Charlie", "IT", 60000),
    (4, "David", "Finance", 70000),
    (5, "Eva", "HR", 45000)
]

columns = ["emp_id", "name", "department", "salary"]

df = spark.createDataFrame(data, columns)

display(df)
df.printSchema()
df.show()
df.count()
it_employees = df.filter(df.department == "IT")
display(it_employees)
display(df.select("name", "salary"))
from pyspark.sql.functions import col
df_bonus = df.withColumn("salary_after_hike", col("salary") * 1.10)
display(df_bonus)
df.groupBy("department").agg(
avg("salary").alias("avg_salary"),
max("salary").alias("max_salary"),
min("salary").alias("min_salary")
).show()
display(
    df.orderBy(df.salary.desc(),df.name.desc())
)
df.createOrReplaceTempView("employees")

display(spark.sql("""
SELECT department,
COUNT(*) AS employee_count,
AVG(salary) AS avg_salary
FROM employees
GROUP BY department
ORDER BY avg_salary DESC
"""))
dept_data = [
    ("IT", "Technology"),
    ("HR", "Human Resources"),
    ("Finance", "Financial Services")
]

dept_df = spark.createDataFrame(
    dept_data,
    ["department", "dept_full_name"]
)

joined_df = df.join(dept_df, "department", "inner")
display(joined_df)
abv5k_employees = df.filter(df.salary > 50000)
display(abv5k_employees)
display(df.filter(col("salary") > 50000))
outer_joined_df = df.join(dept_df, "department", "full_outer")
display(outer_joined_df)

right_joined_df = df.join(dept_df, "department", "right_outer")
display(right_joined_df)
```
